```python
#Recursion method 

def prob_landing_2(x0, x1):
    distance = x1 - x0

    if distance == 0:
        return 1
    if distance == 1:
        return 0.5

    return 0.5 * prob_landing_2(x0, x1 - 1) + \
           0.5 * prob_landing_2(x0, x1 - 2)

x0,x1 = 1,26
print("\n"*4)
print("Using recursion, the probability of landing on square" ,x1, "when starting at square",x0,"is:")
print(prob_landing_2(x0, x1))
print("\n"*4)
        
        
```
