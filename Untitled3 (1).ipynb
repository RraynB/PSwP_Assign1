```python
#Recursion method 

def prob_landing_1(x0, x1):
    distance = x1 - x0

    if distance == 0:
        return 1
    if distance == 1:
        return 0.5

    return 0.5 * prob_landing_1(x0, x1 - 1) + \
           0.5 * prob_landing_1(x0, x1 - 2)

x0,x1 = 1,26
print("\n"*4)
print("Using recursion, the probability of landing on square" ,x1, "when starting at square",x0,"is:")
print(prob_landing_2(x0, x1))
print("\n"*4)
        
        
```

```python
#Brute force "choice method"

import math
def prob_landing_2(x0,x1): 
    distance = x1-x0
    prob = 0
    for i in range(distance // 2 + 1 ):
        additive = math.factorial(distance-i)/(math.factorial(i)*math.factorial(distance-2*i))
        prob += additive*(0.5)**(distance-i)

    return(prob)

x0,x1 = 1,26
print("\n"*4)
print("Using the binomial formula, the probability of landing on square" ,x1, "when starting at square",x0,"is:")
print(prob_landing_2(x0, x1))
print("\n"*4)


        
    
```

```python
import pandas as pd
import matplotlib.pyplot as plt


squares = range(2, 27)

results = []

for n in squares:
    recursion = prob_landing_1(1, n)
    binomial = prob_landing_2(1, n)

    results.append({
        "Final square": n,
        "Recursion ": recursion,
        "Binomial": binomial,
        "Difference": abs(recursion - binomial)
    })

df = pd.DataFrame(results)
df.round(10)

fig, ax = plt.subplots(figsize=(10, 8))
ax.axis("off")

table = ax.table(
    cellText=df.round(10).values,
    colLabels=df.columns,
    cellLoc="center",
    loc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 1.5)

plt.show()
```
